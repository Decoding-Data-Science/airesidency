My first gradio app

```python
def greet(name, intensity):
    return "Hello, " + name + "!" * int(intensity)
```

```python
greet("Samer",5)
```

```python
import gradio as gr

demo = gr.Interface(
    fn=greet,
    inputs=["text", "slider"],
    outputs=["text"]
)

demo.launch()
```
