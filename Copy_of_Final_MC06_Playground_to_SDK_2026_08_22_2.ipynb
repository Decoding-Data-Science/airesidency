## Part B — Now, the same chatbot written directly in code

This time, instead of pointing to a saved prompt, we write the instructions ourselves, right here in Python. This is the pattern you'll use once you're building a real application — the instructions live in your code, not in a separate saved prompt.

Everything else about the call looks almost identical to Part A.

```python
# Our instructions, written directly in code this time
SYSTEM_INSTRUCTIONS = (
    '''
You are an expert Python programming tutorbot. Your objective is to help users learn Python by:
- Explaining concepts in simple, clear language.
- Breaking down problems step-by-step.
- Guiding learners through the process of coding, debugging, and reasoning about code.
- Giving helpful hints before providing solutions, if appropriate.
- Answering questions at any programming level (beginner, intermediate, advanced), adjusting your explanations to the learner’s background where possible.

Before giving a concrete answer, start by reasoning step-by-step or prompting the user to think critically, as appropriate. Only provide concise Python code examples if needed, and explain what the code does before presenting it.

If the user asks a question that could be solved with code, first outline the logical steps to solve it; only then present the code and explanation.

Always ensure your response format is:
1. Step-by-step reasoning and explanation.
2. Python code (if needed), shown with concise inline or block formatting as context demands.
3. Concise summary/conclusion.

Example 1:
User input: "How do I reverse a list in Python?"

Output:
Reasoning: To reverse a list, we can either use slicing or the built-in .reverse() method—let’s break down both methods.
Example code:
my_list = [1, 2, 3, 4]
# Using slicing
reversed_list = my_list[::-1]
# Using .reverse()
my_list.reverse()
Summary: Slicing creates a reversed copy; .reverse() modifies the list in place.

Example 2:
User input: "Write a function to check if a string is a palindrome."

Output:
Reasoning: To check for a palindrome, we want to compare the string to its reverse. If they are equal, it’s a palindrome.
Example code:
def is_palindrome(s):
    return s == s[::-1]
Summary: The function returns True if the input string is a palindrome.

Responses should be concise and use Python best practices. All explanations and code should be tailored to the user’s experience level, as inferred from the question.

Output format: Short paragraphs, followed by necessary Python code with proper comments, ending with a single-sentence summary.

(Reminder: Your primary goal is to provide step-by-step explanations BEFORE solutions. Always adapt your language and details to the user’s apparent experience level.)
    '''
)

response = client.responses.create(
    model="gpt-5.6-terra",
    reasoning={"effort": "medium"},
    instructions=SYSTEM_INSTRUCTIONS,
    input="Hi! Can you help me understand what you can do?"
)

print(response.output_text)
```

**What's `reasoning={"effort": "medium"}`?** We're using a reasoning model, and this one line controls how much the model "thinks" before answering — `"low"` for quick answers, `"medium"` for balanced, `"high"` when the task needs deeper thinking. That's the only setting we need to know for today — keep it on `"medium"` unless we say otherwise.

## Part C — Make it a real, back-and-forth chatbot

Right now, each call answers one message and forgets it ever happened. A real chatbot needs to remember the conversation. We do that by keeping a running list of every message — yours and the bot's — and sending the whole list back each time.

Run the cell below, then type in the box that appears. Type `quit` to stop.

```python
def chat_with_bot():
    print("DDS Chatbot — type 'quit' to exit\n")
    conversation = []

    while True:
        user_input = input("You: ")
        if user_input.strip().lower() == "quit":
            print("Bot: Goodbye!")
            break

        conversation.append({"role": "user", "content": user_input})

        response = client.responses.create(
            model="gpt-5.6-terra",
            reasoning={"effort": "medium"},
            instructions=SYSTEM_INSTRUCTIONS,
            input=conversation
        )

        reply = response.output_text
        print("Bot:", reply, "\n")
        conversation.append({"role": "assistant", "content": reply})

chat_with_bot()
```

Try asking it something, then follow up with a question that only makes sense if it remembers your first message (e.g. "what did I just ask you?"). If it answers correctly, the memory is working — that's because we're sending the *entire* conversation list back each time, not just your latest message.

```python
import gradio as gr

def chatbot_response(message, history):
    conversation = [
        {"role": h["role"], "content": h["content"] if isinstance(h["content"], str) else str(h["content"])}
        for h in history
    ] + [{"role": "user", "content": message}]

    response = client.responses.create(
        model="gpt-5.4",
        reasoning={"effort": "medium"},
        instructions=SYSTEM_INSTRUCTIONS,
        input=conversation
    )
    return response.output_text

gr.ChatInterface(chatbot_response, title="DDS Chatbot").launch()
```
